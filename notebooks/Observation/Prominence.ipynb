## Observation with DST

```python
import scipy.io as sio
%matplotlib inline

filename = '../../data/obs/prom_Hsp_20200621.sav'
d = sio.readsav(filename)
print(d.date_obs)
d.keys()
```

```python
import matplotlib.pyplot as plt   # module for plotting

fig = plt.figure(figsize=(13,4))
ax1 = fig.add_subplot(1,3,1 )
ax1.plot(d.wla2,d.spa2*1e4)
ax1.set_xlabel('wavelength [A]',fontsize=15)
ax1.set_ylabel('$I_{\lambda}$ in $I_{c,disk} \\times 10^{-4}$',fontsize=15)
#ax1.set_ylim(-0.5,1E0)
plt.title(r"H$\alpha$",fontsize=16)
ax2 = fig.add_subplot(1,3,2 )
ax2.plot(d.wlb2,d.spb2*1e4)
ax2.set_xlabel('wavelength [A]',fontsize=15)
plt.title(r"H$\beta$",fontsize=16)
ax3 = fig.add_subplot(1,3,3 )
ax3.plot(d.wlg2,d.spg2*1e4)
ax3.set_xlabel('wavelength [A]',fontsize=15)
plt.title("H$\gamma$",fontsize=16)
plt.show()
```
