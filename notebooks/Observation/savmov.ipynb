# animate img cube saved in IDL save file

```python
import scipy.io as sio
#%matplotlib inline
%matplotlib widget
from ipywidgets import interact
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np

import ipywidgets as widgets
from PIL import Image
```

```python
filename = 'd:/tmp/mov_.sav'
d = sio.readsav(filename)
d.keys()
d.mov_.shape
```

```python

#plt.imshow(d.mov_[0,:,:],origin='lower')
img_cube=d.mov_
nt = 200

fig = plt.figure(figsize=(3,3), dpi=100)
ax = fig.add_axes([0.05,0.05,0.9,0.9])

# you can change cmap or contrast lower/upper limit here
im = ax.imshow(img_cube[0,:,:], cmap="gray",origin='lower')#, vmin=-1, vmax=+1)


def update_img(t=0):
    im.set_data(img_cube[t,:,:])

interact(update_img, t=widgets.IntSlider(min=0, max=nt-1, step=1, value=0))
```

```python
# define a player class
class Player:
    
    def __init__(self, img_cube):
        
        self.img_cube = img_cube
        self.init_widget(img_cube.shape[0])
        
    def init_widget(self, nt):
        
        fig = plt.figure(figsize=(3,3), dpi=100)
        ax = fig.add_axes([0.05,0.05,0.9,0.9])

        # you can change cmap or contrast lower/upper limit here
        self.im = ax.imshow(self.img_cube[0,:,:], cmap="gray",origin='lower')#, vmin=-1, vmax=+1)
        
        # interval sets the time interval in [msec] of autoplay
        play_button=widgets.Play(min=0,max=nt-1,step=1,interval=50,disabled=False)
        slider = widgets.IntSlider(min=0, max=nt-1, step=1, value=0)
        widgets.jslink((play_button,"value"),(slider,"value"))
        widget=widgets.interactive(self.slider_update,t=slider)
        self.out=widgets.Output()
        app=widgets.VBox([widgets.HBox([play_button,slider]),self.out])
        display(app)
    
    def slider_update(self, t):
        
        with self.out:
            self.im.set_data(self.img_cube[t,:,:])
```

```python
# input image cube data, with t axis at the first dimension.
# this player will update images along its first dimension
Player(img_cube);
```
